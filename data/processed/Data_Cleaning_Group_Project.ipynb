Creating bins for age

```python
age_edges = [0, 55, 65, 1000]
age_labels = ['40-55', '55-65', '65+']

# Create the new categorical variable
df['Age_Group'] = pd.cut(df['age'], bins=age_edges, labels=age_labels)
```

Creating bins for time

```python
time_edges = [0, 60, 120, 180, 240, 300]
time_labels = ['0-2 Months', '2-4 Months', '4-6 Months', '6-8 Months', '8+ Months']

# Create the new categorical variable
df['Time_Group'] = pd.cut(df['time'], bins=time_edges, labels=time_labels)
```

Creating bins for EF

```python
EF_edges = [0, 30, 40, 50, 100]
EF_labels = ['Critically Low', 'Very Low', 'Mildly Low', 'Normal']

# Create the new categorical variable
df['EF_Group'] = pd.cut(df['ejection_fraction'], bins=EF_edges, labels=EF_labels)
```

Creating bins for Creatinine levels

```python
Creatinine_edges = [0, 0.7, 1.3, 100]
Creatinine_labels = ['Low', 'Normal', 'High']

# Create the new categorical variable
df['Creatinine_Group'] = pd.cut(df['serum_creatinine'], bins=Creatinine_edges, labels=Creatinine_labels)
```

Creating bins for Sodium levels

```python
Sodium_edges = [0, 120, 124, 134, 145, 1000]
Sodium_labels = ['Severely Low', 'Moderately Low', 'Mildly Low', 'Normal', 'High']

# Create the new categorical variable
df['Sodium_Group'] = pd.cut(df['serum_sodium'], bins=Sodium_edges, labels=Sodium_labels)
```

Creating bins for Platelets

```python
platelets_edges = [0, 150000, 450000, 1000000]
platelets_labels = ['Low', 'Normal', 'High']

# Create the new categorical variable
df['platelets_Group'] = pd.cut(df['platelets'], bins=platelets_edges, labels=platelets_labels)
```

Creating bins for creatinine phosphokinase

```python
CPK_edges = [10, 120, 500, 5000, 1000000]
CPK_labels = ['Normal', 'Mildly high', 'Moderately high', 'Severely high']

# Create the new categorical variable
df['CPK_Group'] = pd.cut(df['creatinine_phosphokinase'], bins=CPK_edges, labels=CPK_labels)
```
